## Objetivo

Aplicar el algoritmo K-Means para agrupar las 24 jurisdicciones de Argentina según su cantidad de viviendas, identificando patrones regionales y segmentando el territorio en categorías de tamaño habitacional.

## 1. Importación de librerías

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

## 2. Definición del conjunto de datos

Se cargan las 24 jurisdicciones con su cantidad de viviendas.

```python
registros = [
    {"jurisdicción": "Ciudad Autónoma de Buenos Aires", "viviendas": 1391258},
    {"jurisdicción": "Buenos Aires", "viviendas": 5970702},
    {"jurisdicción": "Catamarca", "viviendas": 131978},
    {"jurisdicción": "Chaco", "viviendas": 368728},
    {"jurisdicción": "Chubut", "viviendas": 213317},
    {"jurisdicción": "Córdoba", "viviendas": 1378237},
    {"jurisdicción": "Corrientes", "viviendas": 370958},
    {"jurisdicción": "Entre Ríos", "viviendas": 494473},
    {"jurisdicción": "Formosa", "viviendas": 194689},
    {"jurisdicción": "Jujuy", "viviendas": 238141},
    {"jurisdicción": "La Pampa", "viviendas": 140879},
    {"jurisdicción": "La Rioja", "viviendas": 124149},
    {"jurisdicción": "Mendoza", "viviendas": 639467},
    {"jurisdicción": "Misiones", "viviendas": 420101},
    {"jurisdicción": "Neuquén", "viviendas": 254545},
    {"jurisdicción": "Río Negro", "viviendas": 276371},
    {"jurisdicción": "Salta", "viviendas": 404504},
    {"jurisdicción": "San Juan", "viviendas": 241436},
    {"jurisdicción": "San Luis", "viviendas": 182886},
    {"jurisdicción": "Santa Cruz", "viviendas": 118047},
    {"jurisdicción": "Santa Fe", "viviendas": 1273460},
    {"jurisdicción": "Santiago del Estero", "viviendas": 311361},
    {"jurisdicción": "Tierra del Fuego, Antártida e Islas del Atlántico Sur", "viviendas": 65535},
    {"jurisdicción": "Tucumán", "viviendas": 493794}
]

df = pd.DataFrame(registros)
df.head()
```

## 3. Normalización de los datos

Se aplica StandardScaler para evitar que la magnitud de los valores afecte el cálculo de distancias.

```python
X = df[["viviendas"]].values
escalador = StandardScaler()
X_esc = escalador.fit_transform(X)
```

## 4. Determinación de K óptimo (método del codo)

Se calcula la inercia para K entre 1 y 10 y se grafica la curva resultante.

```python
inercias = []
valores_k = range(1, 11)

for k in valores_k:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_esc)
    inercias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(valores_k, inercias, marker="o", color="#2e7d32", linewidth=2)
plt.title("Método del Codo — Selección de K")
plt.xlabel("Número de Clusters (K)")
plt.ylabel("Inercia")
plt.xticks(valores_k)
plt.grid(True, alpha=0.3)
plt.savefig("grafico_codo.png", dpi=150, bbox_inches="tight")
plt.show()
```

La curva muestra una caída pronunciada hasta K=2 y una estabilización a partir de K=3. Se selecciona **K=3** como valor óptimo.

## 5. Entrenamiento del modelo con K=3

```python
modelo = KMeans(n_clusters=3, random_state=42)
df["cluster"] = modelo.fit_predict(X_esc)
df.sort_values("viviendas")
```

## 6. Visualización de los clusters

Se utiliza una paleta de colores verde, naranja y azul para diferenciar los grupos.

```python
paleta = {0: "#2e7d32", 1: "#f57c00", 2: "#1565c0"}

plt.figure(figsize=(11, 6))
for c in range(3):
    grupo = df[df["cluster"] == c]
    plt.scatter(grupo["jurisdicción"], grupo["viviendas"],
                label=f"Grupo {c}", color=paleta[c], s=70, edgecolor="black", linewidth=0.4)

plt.xticks(rotation=90)
plt.xlabel("Jurisdicción")
plt.ylabel("Cantidad de viviendas")
plt.title("Segmentación de Viviendas por Jurisdicción — K-Means")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("grafico_segmentacion.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 7. Conclusiones

- El método del codo permitió identificar **K=3** como el número óptimo de clusters.
- Se obtuvieron tres grupos claramente diferenciados por la cantidad de viviendas.
- Buenos Aires se destaca como caso atípico con más de 5.9 millones de viviendas.
- La normalización fue clave para evitar sesgos por magnitud.
